```python



# ============================================
# 15. Detect Outliers using IQR
# ============================================

print("\n========== OUTLIERS ==========")
numeric_columns = ['open','high','low','close','volume']
outlier_counts = {}

for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    )

    outlier_counts[col] = outliers.sum()

    print("\n---", col, "---")
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", outliers.sum())



```

###  Outliers Handling Decision
**Decision:** We decided **not** to remove or cap the extreme values (outliers) in the stock price and volume columns.
**Reasoning:** In financial datasets, extreme highs or lows represent real-world economic events, market shocks, or major historical growth periods for the stocks (not data entry errors). Deleting or artificially capping them would distort the true historical trends and compromise future model predictions.

```python
import pandas as pd
import numpy as np
import yfinance as yf

# 1. ترتيب البيانات زمنياً لكل سهم بدقة
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df = df.sort_values(by=['Name', 'date']).reset_index(drop=True)

start_date = df['date'].min().strftime('%Y-%m-%d')
end_date = df['date'].max().strftime('%Y-%m-%d')

# سحب بيانات مؤشر السوق العام
market_data = yf.download('^GSPC', start=start_date, end=end_date)
if isinstance(market_data.columns, pd.MultiIndex):
    market_data.columns = market_data.columns.get_level_values(0)
market_data = market_data.reset_index()
market_df = market_data[['Date', 'Close']].rename(columns={'Close': 'Market_Close', 'Date': 'date'})
market_df['date'] = pd.to_datetime(market_df['date'])

df = pd.merge(df, market_df, on='date', how='left')
df['Market_Close'] = df['Market_Close'].ffill()

# 2. المؤشرات الأساسية والفنية الاحترافية
df['Price_Range'] = df['high'] - df['low']
df['Daily_Change'] = df['close'] - df['open']

df['avg_close_10d'] = df.groupby('Name')['close'].transform(lambda s: s.shift(1).rolling(10).mean())
df['trend_10d'] = df['close'] - df['avg_close_10d']

# مؤشر الزخم RSI
def compute_rsi(series, period=14):
    delta = series.diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=period).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=period).mean()
    rs = gain / (loss + 1e-8)
    return 100 - (100 / (1 + rs))

df['RSI'] = df.groupby('Name')['close'].transform(lambda x: compute_rsi(x))

# مؤشر MACD
exp12 = df.groupby('Name')['close'].transform(lambda x: x.ewm(span=12, adjust=False).mean())
exp26 = df.groupby('Name')['close'].transform(lambda x: x.ewm(span=26, adjust=False).mean())
df['MACD'] = exp12 - exp26

# مؤشر Bollinger Bands Width
rolling_mean = df.groupby('Name')['close'].transform(lambda x: x.rolling(20).mean())
rolling_std = df.groupby('Name')['close'].transform(lambda x: x.rolling(20).std())
df['BB_Width'] = (4 * rolling_std) / (rolling_mean + 1e-8)

# مؤشر السوق وصدمة الحجم
df['Market_Return'] = df.groupby('Name')['Market_Close'].pct_change()
df['Volume_SMA_10'] = df.groupby('Name')['volume'].transform(lambda x: x.rolling(window=10).mean())
df['Volume_Ratio'] = df['volume'] / df['Volume_SMA_10']
df['Return_5d'] = df.groupby('Name')['close'].pct_change(periods=5)
df['Price_to_High_20d'] = df['close'] / df.groupby('Name')['high'].transform(lambda x: x.rolling(20).max())
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)
# 3. **إضافة جديدة:** عوائد الأيام السابقة (Lag Returns) ليعرف المودل الزخم
df['Return_1d'] = df.groupby('Name')['close'].pct_change()
df['Return_2d'] = df.groupby('Name')['close'].pct_change(periods=2)

# 4. **التعديل الجذري للـ Target:** العائد على مدار 3 أيام قادمة ممهد لتقليل التشويش
df['Target_Return'] = (df.groupby('Name')['close'].shift(-3) / df['close'] - 1).rolling(window=2, min_periods=1).mean()

# تنظيف القيم الفارغة
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)

print("تم تطبيق التعديلات والتحسينات بنجاح! الشكل النهائي للداتا:", df.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

# تحويل أسماء الشركات إلى أرقام باستخدام Label Encoding
le = LabelEncoder()
df['Name_Encoded'] = le.fit_transform(df['Name'])

# استعراض الأعمدة بعد التشفير
print(df[['Name', 'Name_Encoded']].head())
```

```python
from sklearn.model_selection import TimeSeriesSplit
from sklearn.preprocessing import StandardScaler
from catboost import CatBoostRegressor
from sklearn.ensemble import RandomForestRegressor, VotingRegressor
from xgboost import XGBRegressor  # <--- التصحيح هنا
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. هندسة ميزات متقدمة إضافية
high_low = df['high'] - df['low']
high_close = np.abs(df['high'] - df['close'].shift())
low_close = np.abs(df['low'] - df['close'].shift())
tr = pd.concat([high_low, high_close, low_close], axis=1).max(axis=1)
df['ATR'] = tr.rolling(window=14).mean()

df['Price_Slope'] = df.groupby('Name')['close'].transform(lambda x: x.diff(5) / 5)

df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)

features = [
    'Name_Encoded', 'open', 'high', 'low', 'close', 'volume',
    'Price_Range', 'Daily_Change', 'avg_close_10d', 'trend_10d',
    'RSI', 'MACD', 'BB_Width', 'Market_Return', 'Volume_Ratio',
    'Return_1d', 'Return_2d', 'Return_5d', 'Price_to_High_20d',
    'ATR', 'Price_Slope'
]

X = df[features]
y = df['Target_Return']

# 2. التقسيم الزمني
tscv = TimeSeriesSplit(n_splits=5)
for train_index, test_index in tscv.split(X):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

# 3. التحجيم
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. النماذج الفردية
rf_model = RandomForestRegressor(n_estimators=300, max_depth=14, min_samples_split=4, random_state=42, n_jobs=-1)
cat_model = CatBoostRegressor(iterations=400, learning_rate=0.03, depth=6, verbose=0, random_seed=42)
xgb_model = XGBRegressor(n_estimators=250, learning_rate=0.03, max_depth=5, random_state=42, n_jobs=-1)

# 5. الـ Ensemble
ensemble_model = VotingRegressor(estimators=[
    ('rf', rf_model),
    ('cat', cat_model),
    ('xgb', xgb_model)
])

models = {
    "CatBoost": cat_model,
    "Random Forest": rf_model,
    "XGBoost": xgb_model,
    "Ensemble (Voting)": ensemble_model,
    "Ridge Regression": Ridge(alpha=1.0)
}

results = []
trained_predictions = {}

print("Training and evaluating models...")
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    trained_predictions[name] = y_pred

    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))

    results.append({"Model": name, "MAE": mae, "RMSE": rmse})
    print(f"-> Finished: {name}")

# 6. جدول المقارنة
results_df = pd.DataFrame(results).sort_values(by="MAE")
print("\n" + "="*50)
print(" Ultimate Ensembled Model Performance Comparison:")
print("="*50)
print(results_df.to_string(index=False))

# 7. أفضل نموذج للرسم
best_model_name = results_df.iloc[0]["Model"]
best_predictions = trained_predictions[best_model_name]

print(f"\nThe Best Performing Model is: {best_model_name}")

# 8. الرسم البياني
plt.figure(figsize=(12, 6))
plt.plot(y_test.values[:150], label='Actual 3-Day Return', color='blue', linewidth=1.5, alpha=0.7)
plt.plot(best_predictions[:150], label=f'Predicted Return ({best_model_name})', color='orange', linestyle='--', linewidth=2)

plt.title(f'Ultimate Performance - Actual vs Predicted ({best_model_name})', fontsize=14, fontweight='bold')
plt.xlabel('Time Steps', fontsize=12)
plt.ylabel('Target Return (3-Day)', fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
